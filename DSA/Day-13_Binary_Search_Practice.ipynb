```python

def binary_search(nums,target):
    
    n=len(nums)
    low=0
    high=n-1
    
    
    while low<=high:
        mid=(low+high)//2
        if nums[mid]==target:
            return target
        elif  nums[mid]<target :
            low=mid+1
            
        else:
            high=mid-1
    return -1
```

```python
nums=[2,4,6,7,8,9,10,12,13]
binary_search(nums,7)
```
