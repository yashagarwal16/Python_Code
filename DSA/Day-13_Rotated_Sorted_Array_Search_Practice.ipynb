```python
nums=[10,11,12,12,13,13,13,1,2,3,4]
def sor_arr(nums):
    n=len(nums)
    target=13
    low=-1
    high=-1
    
    while low<=high:
        mid=(high-low)//2
        if nums[mid]==target:
            return True
        elif nums[low]<=nums[mid]:
            if nums[low] <= target < nums[mid]:
                    high = mid - 1
            else:
                low=mid-1
        else:
                if nums[mid] < target <= nums[high]:
                    low = mid + 1
                else:
                    high = mid - 1
    
    return False
```
